```python
years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
yearly_datasets = {}

for year in years: 
    file_path = DATA_DIR / f"{year}.xlsx"
    
    print(f"Loading {file_path.name} ...")
   
    yearly_datasets[year] = pd.read_excel(file_path)
    
    n_rows, n_columns = yearly_datasets[year].shape
    print(f"{year}: {n_rows} rows and {n_columns} columns")
```

Before joining the yearly files, we need to know whether they contain the same columns.

Suppose 2025 contains:

Winner
Loser
WRank
LRank
AvgW
AvgL

but 2023 does not contain one of the odds columns.

If we combine the files without checking, we may unexpectedly create many missing values.

set(...)

A set is a collection of unique elements.

Sets allow us to compare groups of column names.

reference_columns - current_columns

This finds columns that are present in the 2025 dataset but missing from the current dataset.

current_columns - reference_columns

This finds columns present in the current dataset but absent from 2025.

```python
reference_columns = set(yearly_datasets[2025].columns)

for year, dataset in yearly_datasets.items(): 
    current_columns = set(dataset.columns)
    
    missing_compared_to_2025 = reference_columns - current_columns
    additional_compared_to_2025 = current_columns - reference_columns
    
    print(f"\nYear: {year}")
    print("Columns missing compared with 2025: ")
    print(sorted(missing_compared_to_2025))
    
    print("Additional columns compared to 2025: ")
    print(sorted(additional_compared_to_2025))
```
